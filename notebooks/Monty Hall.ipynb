## Monty Hall

```python
import numpy as np

def monty_hall_numpy(n_trials=1_000_000):
    # Step 1: Randomly place the car behind doors 1, 2, or 3
    car = np.random.randint(1, 4, size=n_trials)

    # Step 2: Player randomly picks a door
    player = np.random.randint(1, 4, size=n_trials)

    # Step 3: Monty opens a goat door
    # Monty cannot open the player's door or the car door.
    monty = np.zeros(n_trials, dtype=int)

    # Pre-calculate indices for cases
    player_picked_car_idx = (player == car)

    # Case 1: Player picked the car (player == car)
    # Monty must open one of the two other doors (not the player's door).
    # Create a mapping for possible Monty choices if player picked the car.
    # E.g., if player picked 1, Monty can open 2 or 3.
    # Indices are 0-based for array: player-1
    # monty_choices_map[0] -> [2, 3] (for player == 1)
    # monty_choices_map[1] -> [1, 3] (for player == 2)
    # monty_choices_map[2] -> [1, 2] (for player == 3)
    monty_choices_map = np.array([[2, 3], [1, 3], [1, 2]])

    # Generate random indices (0 or 1) to pick one of the two options for each trial
    # where player == car
    random_choice_for_monty = np.random.randint(0, 2, size=n_trials)

    # Assign Monty's choice for trials where player == car
    monty[player_picked_car_idx] = monty_choices_map[
        player[player_picked_car_idx] - 1,
        random_choice_for_monty[player_picked_car_idx]
    ]

    # Case 2: Player picked a goat (player != car)
    # Monty must open the *other* goat door (the one that is not player's choice and not car's location).
    # Since doors are 1, 2, 3, and their sum is 6, the third door is 6 - door1 - door2.
    monty[~player_picked_car_idx] = 6 - player[~player_picked_car_idx] - car[~player_picked_car_idx]

    # Step 4: Determine the remaining closed door (the switch door)
    # The switch door is the one that is not the player's initial choice and not the door Monty opened.
    # Again, sum of doors is 6.
    switch = 6 - player - monty

    # Step 5: Count wins
    stay_wins = np.sum(player == car)
    switch_wins = np.sum(switch == car)

    return stay_wins, switch_wins

stay, switch = monty_hall_numpy()
print(f"Stay win rate:   {stay / 1_000_000:.3f}")
print(f"Switch win rate: {switch / 1_000_000:.3f}")
```
